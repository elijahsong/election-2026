# 2024 Election Night — Consolidated Research Thread

**Historical reference for the experimental 2026 Senate nowcast**  
**Scope:** November 5–6, 2024 and subsequent counting.  
**Purpose:** Operational reconnaissance—decide which states/counties to watch and when, **not** model training or statistical validation.

This notebook preserves the prior research summary of **AP and Decision Desk HQ (DDHQ) race calls, vote-counting speed, Kalshi market evidence, county reporting surprises, and contemporary news accounts**. Values marked unverified or approximate should not be treated as exact archival measurements.

## 1. Key observations

- **Florida** reported exceptionally quickly and is the leading candidate for the first manually entered 2026 nowcast signal.
- **Georgia** could be informative early, but delayed large metro Atlanta counties—including Gwinnett—distorted the interim statewide picture.
- **North Carolina and Virginia** were useful potential cross-checks, but reporting order made raw partial margins misleading.
- **Texas** may produce its own returns soon enough that the Florida-to-Texas inference window is brief.

## 2. Contemporary reporting surprises

### Georgia — Gwinnett and metro Atlanta

| Time (EST), Nov 5 | Contemporary report |
|---|---|
| ~8:15 PM | Gwinnett officials discussed a write-in ballot adjudication/software delay; roughly 250,000 early/absentee votes awaited transmission, according to CNN reporting. |
| ~9:00 PM | FOX 5 Atlanta warned substantial Democratic-leaning metro Atlanta returns, including Fulton, DeKalb, Cobb, and Clayton, were still expected. |
| ~9:50 PM | FOX 5 reported Gwinnett reporting was still only in the teens (percentage reporting). |
| ~11:15 PM | Gwinnett officials said the software-related delay had been resolved. |

Fulton County also experienced disruptions related to bomb threats at polling locations. The key issue was **which votes remained**, not simply the statewide fraction counted.

Sources: [CNN election-night transcript](https://transcripts.cnn.com/show/se/date/2024-11-05/segment/34), [FOX 5 Atlanta live updates](https://www.fox5atlanta.com/news/election-day-live-updates-polls-now-open-state-georgia).

### North Carolina — major counties absent at 9 PM

Contemporaneous WUNC reporting indicated Wake and Mecklenburg had not reported Election Day votes, while Guilford and Gaston had not reported any votes at all. Long lines in some areas also delayed returns. The 2024 early-vote reporting process was affected by procedural changes.

Source: [WUNC 2024 live updates](https://www.wunc.org/live-updates/2024-election-day-north-carolina-voting-results-live).

### Virginia — misleading interim statewide lead

An earlier report placed Virginia at approximately **57% counted around 8:46 PM**, with Trump narrowly leading the reported count despite Harris ultimately winning by roughly 5.8 percentage points. **The precise interim snapshot remains to be independently verified.** This is a useful qualitative illustration of reporting-order bias.

## 6. Kalshi market timing

Proposed **operational** measure: the first time the eventual winner's contract reaches **98% or 99% implied probability for at least five consecutive minutes**. This is *time to sustained market certainty*, not an AP/DDHQ race call, not an official count-completion measure, and not a predictor for the pre-election model.

### Contemporaneous national presidential market observations

| Time (EST) | Approximate national Trump contract probability | Qualification |
|---|---:|---|
| Nov 5, early evening | 58–60% | Rough contemporaneous observation |
| By ~11:00 PM | ~90% | Approximate reported level |
| By Nov 6 1:01 AM | ~95% | Reported by Fortune; **not** the exact first crossing |
| By Nov 6 1:38 AM | ~99% | Reported by Forbes; **not** the exact first crossing |

**State-specific 98%/99% crossing timestamps remain unverified** and should not be populated by guessing. Historical Kalshi trades and one-minute candlesticks can potentially recover them.

Official API references:

- [Kalshi historical data guide](https://docs.kalshi.com/getting_started/historical_data)
- [Historical market candlesticks](https://docs.kalshi.com/api-reference/historical/get-historical-market-candlesticks)
- [Historical trades](https://docs.kalshi.com/api-reference/historical/get-historical-trades)

Implementation notes: record pre-poll-close thresholds separately; distinguish stale last-trade prices from active quotes; missing candles do not prove five continuous minutes above threshold.

## 8. Proposed 2026 monitoring sequence (EST)

| Window | Plan |
|---|---|
| 7:00–8:00 PM | Monitor Georgia and Virginia; check which metro counties have reported; avoid strong conclusions from raw margins. |
| 8:00–8:30 PM | Focus on Florida statewide plus Miami-Dade, Broward, Pinellas, Orange, and Duval. Compare to frozen expectations. |
| 8:30–9:00 PM | Consider first provisional Florida-to-Texas adjustment, but check whether direct Texas returns are already informative. |
| 9:00–10:00 PM | Cross-check with Georgia, Virginia, North Carolina, and Ohio; watch large outstanding counties. |
| After 10:00 PM | Prefer direct Texas observations and other state-specific returns over indirect extrapolation. |

A simple discretionary live-update equation:

$$
\widehat M_{\mathrm{TX,live}}=\widehat M_{\mathrm{TX,pre}}+\lambda_{\mathrm{TX}}\widehat N_{\mathrm{live}}
$$

Here $\widehat N_{\mathrm{live}}$ is a **judgment-adjusted transferable surprise**, not a mechanically observed national shift. Reporting percentages and race-call timing guide *when and what to watch*, not the value of the transfer coefficient.
